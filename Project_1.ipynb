```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
mdf = pd.read_csv('linked_mortality_file_1999_2000.csv') # Load mortality file
print( mdf.head() )
gdf = pd.read_sas("DEMO.xpt", format="xport") # Load demographics file
print( gdf.head() )
df = gdf.merge(mdf, on="SEQN", how="inner")
```

```python
cdf = pd.read_csv("SSCAFE_A.csv")
print(cdf.head())

df2 = cdf.merge(df, on="SEQN", how="inner")
print(df2.head())
df2.to_csv("merged_mortality_caffine.csv", index=False)
```
