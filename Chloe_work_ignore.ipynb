```python
# linear regression
from sklearn.linear_model import LinearRegression
tdf = new_df.loc[:, ['MORTSTAT','SSMX4', 'SSMX5', "SSMX6", "SSMX7"]].dropna()
y = tdf['MORTSTAT']
X_num = tdf.loc[ :, ['SSMX4', 'SSMX5', 'SSMX6', 'SSMX7']]
model = LinearRegression()
model = model.fit(X_num, y)

print(f'Model intercept: {model.intercept_}')
pd.DataFrame({'variable':model.feature_names_in_, 'coefficient':model.coef_ })
```

```python
y_hat = model.predict(X_num)
sns.scatterplot(x = y, y = y_hat)
```

```python
model = LinearRegression()
model = model.fit(X_num.reshape(-1,1), y)
y_hat = model.predict(X_num.reshape(-1,1))

plt.scatter(X_num,y,label='Data')
plt.plot(X_num,y_hat,label='Regression Line')
plt.xlabel("X_num")
plt.ylabel("Y")
plt.legend(loc='lower right')
```
